#Chatbot vs AI Agent

#Components of an AI Agent

```python
# Goal: The exact agenda that the system is trying to solve/achieve
# Input: giving specific information to the Agent
# Planning: breaking the goal into multiple steps
# Tools: The services that the Agent/system can use
# Observation: Feedback/information received after performing an action
# Final response: the output/result shared with the user

```

```python
# Travel Agent:

# Goal - prepare a trip plan
# Input - source, destination, dates range, budget
# Planning - Food, Places, Events/Activities
# Tools - Browser, searching, websites, etc
# Observation - Price of the tickets, etc
# Final response - Itenary shared to the customers
```

#Planning Strategies:

```python
# 1- Step-by-Step planning:
# after completing a particular step, then only check ahead of it.

# 2- Plan-first Approach:
# - create the entire rough plan before performing the actions
# - decide the flight, hotel, food, etc

# 3- Replanning
# - change/update the plan when something unexpected happens.
```

```python
def calculate_total(price, quantity):
    return price * quantity
```

```python
# price, discount_perc --> discounted_price


def calculate_discount(price, discount_percent):
    discount = (price * discount_percent) / 100
    final_price = price - discount

    return final_price
```

```python
def check_stock(product_name):
    stock = {
        "notebook": 50,
        "pen": 100,
        "bag": 12
    }


    return stock.get(product_name, 0)
```

```python
print("Total cost:", calculate_total(40, 5))

print("Price after discount:", calculate_discount(1000, 15))

print("Notebook stock:", check_stock("notebook"))

print("Bag stock:", check_stock("bag"))
```

```python
print("Phone stock:", check_stock("Phone"))
```

```python
def shopping_agent(task):
    task = task.lower()

    if "total" in task:
        print("Agent decision: Use the total-cost tool")

        result = calculate_total(40, 5)

        return f"The total cost is ₹{result}."

    elif "discount" in task:
        print("Agent decision: Use the discount tool")

        result = calculate_discount(1000, 15)

        return f"The final price after discount is ₹{result}."

    elif "stock" in task:
        print("Agent decision: Use the inventory tool")

        result = check_stock("notebook")

        return f"There are {result} notebooks in stock."

    else:
        return "I cannot handle this task yet."


print(shopping_agent("Calculate the total cost"))
print()
print(shopping_agent("Calculate the discount"))
print()
print(shopping_agent("Check the stock"))
```
