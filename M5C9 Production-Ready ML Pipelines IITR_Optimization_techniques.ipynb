#Experimentation vs Production-Ready

```python
# Experimentation phase:

# Dataset --> Train model --> Check the accuracy --> Prediction



# Production-Ready:
# Dataset --> Cleaned & Preprocessed --> ML Model --> Prediction --> Evaluate -->Deploy--> Monitoring
```

```python
# What makes a model production-ready?

# 1. Accurate --> should make useful and correct predictions
# 2. Reliable --> should not randomly fail
# 3. Re-producible --> we should be able to reproduce the same result using the same data/code/settings
# 4. Fast enough --> should not take unreasonable long time to give outcomes
# 5. Maintainable --> other devlopers should be able to understand and update the code.
# 6. Consitent --> training  and predictions should use the same preprocessing steps
# 7. Monitorable --> should be able to detect when the model performance is reducing
```

#A Usual ML Pipeline

```python
# 1. Problem Definition
# 2. Collect the data
# 3. Data Cleaning and Pre-Processing
# 4. EDA & Feature Engineering
# 5. Train the Model
# 6. Make predictions
# 7. Evaluate the model
# 8. Optimize the model
# 9. Save and Deploy
# 7. Monitor and Retrain
```

```python
# Customer churn prediction:

# Features:
# 1-age on the platform
# 2 - number of support tickets/complaints raised
# 3 - Avg ticket size per month
# 4 - no of login-s and purchases over the last month


# Target:
# Churn --> Yes/No
```

```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing()
print(data)
```

```python
x = data.data
y = data.target
```

```python
print(x[:5])
```

```python
print(y[:5])
```

```python
data.feature_names
```

```python
# Train-Test Split:

# Total data --> split into training data and test data

# Training data --> Teach the model to learn the past patterns
# Test data --> to test how the model performs on the unseen data



```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42 )
```

```python
x_train
```

```python
# test_size = 0.2


# test_data = 20%
# training data = 80%
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import RandomForestRegressor

model = RandomForestClassifier(n_estimators= 100, random_state= 40)
model2= RandomForestRegressor()
```

```python
import joblib

model2.fit(x_train, y_train)
```

```python
joblib.dump(model2, "model2.pkl")
```

```python
loaded_model = joblib.load("model2.pkl")
```

```python
prediction = loaded_model.predict(x_test)
print(prediction)
```
