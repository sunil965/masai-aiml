#Ensemble Learning


```python
# its all about combining multiple models to produce a better prediction than we might get from a single model
```

```python
# data --> one model --> prediction
```

```python
# data
#   model1
#   model2
#   model3

#   .
#   .
#   .
#   modeln

# combine predictions of all the models  --> final prediction
```

```python
# Yes/No

# Model1 --> Yes
# Model2 --> No
# Model3 --> No
# Model4 --> Yes
# Model5 --> Yes

# Yes --> 3
# No --> 2

# Final Conclusion --> Yes (Majority)
```

```python
# Types of Ensemble Learning:

# 1. Bagging --> Random Forest
# 2. Boosting --> Gradient Boosting


```

```python
# Bagging --> Models are trained independently/in parallel

# Boosting --> Models are trained sequentially, so later models are trying to imporve the overall predictions
# as they are learning from the past models.
```

#Bagging

```python
# Bagging --> Bootstrap Aggregating

# Bootstrap:
# creating multiple training datasets by sampling form the original dataset with replacement

# Aggregating:
# Combining the predicitons of all the models


# Original Dataset:
# ABCDEF

# Sample1: ACFEAC
# Sample2: CFFEFD
# Sample3L DFEFAA


# Imp points: sampling with replacement

# 1- Some observations appear multiple times
# 2- Some observations dont appear even once
```

#Random forest --> Ensemble of multiple decision trees

```python
# # if 1 model but trained at a very large depth --> it will memorise the training data --> overfitting/variance

# 2 ways to introduce randomness
# 1. Bootstrap samples

# Tree1 --> sample1
# tree2 --> sample2
# tree3 --> sample3


# 2. Random feature selection
# eg:
# F1
# F2
# F3
# F4
# F5

# Tree1 --> F1, F3, F4
# Tree2 --> F2, F3, F5
# Tree3 --> F1, F3, F5
# This way trees become more diverse



```

#Random Forest for Classfication

```python
from sklearn.datasets import load_iris

data = load_iris()

x = data.data
y = data.target

x.shape
```

```python
y.shape
```

```python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(random_state = 42)
tree.fit(x,y)
```

```python
preds = tree.predict(x)
print(preds[:60])
```

```python
from sklearn.ensemble import RandomForestClassifier

tree = RandomForestClassifier(random_state = 42, n_estimators = 10)
tree.fit(x,y)
preds = tree.predict(x)
print(preds[:10])

#n_estimators --> the number of decision trees
```

```python
# More trees --> more computation
# but after some point, additional trees will just add very little improvement
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state = 42)
```

```python
rf_model = RandomForestClassifier(n_estimators=50, random_state = 42)

rf_model.fit(x_train, y_train)
```

```python
y_pred = rf_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```

#Random Forest Regression

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators = 100, random_state = 42)
rf.fit(x_train, y_train)
predictions = rf.predict(x_test)
```

#Boosting


---



```python
# # here the models are built sequentially
# # here the individual models are small decision trees --> that are usually called weak learners

# eg:

# actual value = 100
# predicted = 70

# error1 = 100-70 = 30

# second tree will learn to correct this part of the error
# it contributes 15
# 70+15 = 90

# error2 = 100-90 = 10

# third tree contributes 8
# 90+8 = 98
# error3 = 100-98 = 2

# this is the simplified intuition
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_model = GradientBoostingClassifier(n_estimators = 50, random_state = 42)

gb_model.fit(x_train, y_train)
y_pred = gb_model.predict(x_test)

print(y_pred[:10])
```

```python

accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```
