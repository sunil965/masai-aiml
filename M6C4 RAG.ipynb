#RAG - Retrieval Augmented Generation

```python
# Retrieval: fetching the relevant information
# Augmented: adding information to the prompt/context
# Generation: Asking AI model to generate answer/response using the provided information
```

```python
# Question --> Retrieve relevant information --> add information to the prompt --> LLM --> Response/answer


# Eg: Question --> search textbook --> find relevant page --> read information --> generate answer to the question --> response
```

```python
# User question --> Search the knowledge base --> Retrieve the relevant information --> give that information to the LLM --> generate answer
```

```python
# #Chunking:
# if we have a very heavy document.
# it is not a good idea to pass that document with every query
# as it leads to:
# 1- it is inefficient to every time pass such a huge information
# 2- it can exceed the context limit.
# 3- costly

# Instead, RAG systems breaks down such a document into smaller pieces called chunks.

company policy is of 500+ pages and user is asking query:
"How many paid leaves will I get in a year?"
the query is specific to a particular section instead of unnecessarily scanning the whole document

Large document --> smaller chunks --> search within only the relevant chunks

```

```python
text = """ Employees receive 18 paid leaves per year.

Employees must apply for leave at least 3 working days in advance.

Emergency leave may be requested without prior notice.
"""
```

```python
sentences = text.split(".")
sentences
```

```python
sentences = text.split("\n\n")
sentences
```

```python
i = 1
for s in sentences:
  print("Chunk number:", i, s)
  i += 1
```
