```python
# MAE --> Mean Absolute Error
# MSE --> Mean Squared Error
# RMSE --> Root Mean Squared Error
```

#MAE --> Mean Absolute Error

```python
# | Actual | Predicted |
# | -----: | --------: |
# |    100 |       110 |
# |    200 |       180 |
# |    300 |       330 |

# abs(100-110) = 10
# abs(200-180) = 20
# abs(300-330) = 30


# mean absolute error: (10+20+30)/3 = 60/3 = 20
```

```python
actual = [100,200,300]
predicted = [110, 180, 330]

n = len(actual)
sm = 0

for i in range(n):
  abs_error = abs(actual[i] - predicted[i])
  sm += abs_error

mae = sm/n

print(mae)
```

```python
from sklearn.metrics import mean_absolute_error

actual = [100,200,300]
predicted = [110, 180, 330]

mae = mean_absolute_error(actual, predicted)
print(mae)
```

```python
# | Actual | Predicted |
# | -----: | --------: |
# |    100 |       110 |
# |    200 |       180 |
# |    300 |       330 |

# squared errors = (100-110)^2 + (200-180)^2 + (300-330)^2  = 10^2 + 20^2 + 30^2 = 100 + 400 + 900 = 1400

# mse = 1400/3 = 466.67
```

```python
actual = [100,200,300]
predicted = [110, 180, 330]
n = len(actual)
sm = 0

for i in range(n):
  sm += (actual[i] - predicted[i])**2

mse = sm/n
print(mse)
```

```python
from sklearn.metrics import mean_squared_error

actual = [100,200,300]
predicted = [110, 180, 330]

mse = mean_squared_error(actual, predicted)

print(mse)
```

```python
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(actual, predicted)

print(rmse)
```

```python
# r2_score = how much variation in the target is our model able to explain?
```

```python
# r2_score = 0.9 ==> means the model is able to explain 90% of the variation in the target
```

```python
from sklearn.metrics import r2_score
actual = [100,200,300]
predicted = [110, 180, 330]

r2 = r2_score(actual, predicted)
print(r2)
```

```python
# mae, mse, rmse --> lesser the better
# r2 --> more the better
```

```python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()

x = data.data
y = data.target
```

```python
x[:5]
```

```python
y[:5]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import precision_score, recall_score, accuracy_score, f1_score

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 score:", f1_score(y_test, y_pred))
```
