#Classification Evaluation Metrics

```python
import pandas as pd

data = {
    "Student": [
        "Rahul", "Priya", "Aman", "Neha", "Karan",
        "Sneha", "Rohit", "Anjali", "Vikas", "Pooja"
    ],

    "Actual": [
        1, 1, 0, 1, 0,
        1, 0, 0, 1, 0
    ],

    "Predicted": [
        1, 1, 1, 1, 0,
        0, 0, 1, 1, 0
    ]
}

df = pd.DataFrame(data)

df

#1 -> placed, 0-> not placed
```

#Confusion Matrix

```python
#True Positives --> TP --> The model correctly predicts the positive class
# TP = 1 ===> Rahul, Priya, Neha, Vikas
```

```python
#True Negative --> TN --> The model correctly predicts the negative class
# TN = 1 ===> Karan, Rohit, Pooja
```

```python
#False Positive --> FP --> the model incorrectly predicts it to be a positive class --> False Alarm
# FP = 1 ==> Aman, Anjali
```

```python
#False Negative --> FN --> the model incorrectly predicts it to be a negative class
#FN = 1 ---> Sneha
```

```python
tp = ((df["Actual"] == 1) & (df["Predicted"] == 1)).sum()
print("True positives:", tp)
```

```python
fp = ((df["Actual"] == 0) & (df["Predicted"] == 1)).sum()
print("False positives:", fp)
```

```python
tn = ((df["Actual"] == 0) & (df["Predicted"] == 0)).sum()
print("True Negatives:", tn)
```

```python
fn = ((df["Actual"] == 1) & (df["Predicted"] == 0)).sum()
print("False Negatives:", fn)
```

```python
#                     Predicted
#                   0          1
#               -------------------
# Actual     0   | TN = 3   | FP = 2 |
#            1   | FN = 1   | TP = 4 |
#               -------------------
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(df["Actual"], df["Predicted"])

print(cm)
```

#Accuracy

```python
# Accuracy --> Out of all the predictions, how many are actually correct?
```

```python
# correct --> TP + TN
# total --> TP + TN + FP + FN

# accuracy = correct/total = (TP + TN)/ (TP + TN + FP + FN)
```

```python
accuracy = 7/10
print(accuracy)
```

```python
accuracy = (tp+tn)/(tp+tn+fp+fn)
print(accuracy)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(df["Actual"], df["Predicted"])
print(accuracy)
```

```python
# model 95% accuracy
```

```python
# suppose: 100 students, only 5 got placed.

# model --> always predicts the student to be unplaced.

# accuracy = 95/100
```

#Precision

```python
# precision --> When the model predicts positive, how often is it actually positive?

# actual positive = tp
# total = tp + fp

precision = tp/(tp+fp)
print(precision)
```

```python
from sklearn.metrics import precision_score

precision = precision_score(df["Actual"], df["Predicted"])
print(precision)
```

```python
# model --> these 10 students will get placed.
# but actually, 7 -> placed, 3 -> unplaced

tp1 = 7
fp1 = 3
precision1 = tp1/(tp1 + fp1)
print(precision1)
```

```python
#When is precision useful?

# When the cost of false positive is high, then we prefer precision



# spam detection, spam = 1, not spam = 0

# false positive --> predicting a correct email to be spam

# eg: Loan approaval, positive = yes, negative = no
# false positive -> so approving the loan to a very high default chance person
```

#Recall

```python
# Recall = out of all the actual positives, how many did the model predicted to be positive?
```

```python
# actual positives = tp + fn
print(tp,tn,fp,fn)
recall = tp/(tp + fn)
print(recall)
```

```python
from sklearn.metrics import recall_score

recall = recall_score(df["Actual"], df["Predicted"])
print(recall)
```

```python
#When is recall useful?

# when missing an actual positive case is expensive

# Cancer detection ==> declaring a cancer patient as no cancer can be expensive.

#eg2: Fraud detection --> missing a fraud by detecting it to be a non fraud can be expensive
```

#F1 score

```python
f1_score = (2*precision*recall)/(precision + recall)
print(f1_score)
```

```python
from sklearn.metrics import f1_score

f1 = f1_score(df["Actual"], df["Predicted"])
print(f1)
```
