```python
# Ridge Regression --> it is based on Linear Regression but adds a pentalty for very large coefficients.

# Linear Regression + Penalty for large coefficients = Ridge Regression

# this penalty is controlled by a hyperparameter --> alpha
```

```python
# so basically, alpha is controlling how strongly do we need to penalize the large coefficients
# small alpha --> very little penalty --> behaves very similar to ordinary Linear Regression
# large alpha --> strong penalty --> coefficients are moved to a very smaller values.

```

#Grid Search

```python
# # grid search systematically tries every combination of hyperparameter values that we provide.

# suppose, we have only 1 hyperparamter:
# alpha
# 0.001
# 0.01
# 0.05
# 0.07

# grid search will check for all the 4 different values of alpha
```

```python
# Suppose we have 2 different hyperparameters:

# A
# a1
# a2
# a3


# B
# b1
# b2

# all the possible combinations that will be checked by grid search are:

# a1-b1
# a1-b2
# a2-b1
# a2-b2
# a3-b1
# a3-b2

# total of 3*2 = 6 combinations
```

```python
#Randomized Search: it selects combinations from the search space randomly instead of trying every combination
# Hence it can save us a lot of unnecessary computation

# if a total of 100 different combinations:

#   grid search: tries all 100
#   randomized search might randomly try 10 out of it
```

```python
eg:

suppose 100 different dishes in a restaurant and you want to find the one which you will like:

grid search: try all 100 dishes
problem:
1. time taking -> slow
2. Very expensive


Randomized search:
randomly choose: 10 dishes and try them and you might find something which is excellent without trying all 100 dishes


```
