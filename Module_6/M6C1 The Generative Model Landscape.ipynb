```python
# Tokens:

# LLMs generally do not process an entire sentence as one giant text/object.
# Text is converted into smaller pieces called tokens.

# Eg:

# Text: "I love Programming"

# "I"
# "love"
# "Programming"

# Tokenization depends on the tokenizer --> it need not always be one word -->
# a long word can also be broken down into multiple tokens


# Text --> Tokenizer --> Tokens --> Model --> Predicted tokens --> Text
```

```python
text = "I like to code"
```

```python
tokens = text.split()
```
