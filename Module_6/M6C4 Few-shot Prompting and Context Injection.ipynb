```python
# Types of prompting:
# 1- Zero shot
# 2- One shot
# 3 - Few shot
```

```python
# instruction + examples + new input --> LLM --> Output
```

```python
1- zero shot --> no example provided.
2 - one shot --> 1 example is provided
3 - few shot --> a small number (>1) example is provided.
```

```python
zero-shot prompting:

eg:
1. Please help me with the below mathematical calculation.
2. Please write a code to count the number of digits in the given number.


One-shot prompting:
eg: 1- Write an email to help me apply for a leave. Use this sample email as an example to understand my tone.
2 - I need help with writing a Linkedin post. This is a sample post which gathered a good amount of interaction from the audience.
Please help me create a similar post.


Few shot prompting:
Categorise the given email as spam or not spam:

Eg1: Consider this email to understand what is spam:

Eg2: Consider this email to understand what is not spam:



Sentiment analysis:
3 classes: happy, sad, neutral.
1 eg for each: 3 examples minimum.

image classification:
1- dog
2 - cat
3 - rabbit

```

```python
# Why do we need examples?

# 1-> to reduce down the ambiguity.

# Suppose we have a sentence and we need to convert it into a "professional" language.

# -> the word professional is ambiguous, once we share an eample, there will be better clarity on what it actually means for reference.


# 2-> to convey different elements of the prompt

# 1 eg to convey the output format
# another eg to convey the task in detailed manner
# another eg to convey the dos and donts


# 3-> An example can also be an instruction.

# eg: This is a tweet that should be categorised as "Happy" or "Positive"
```

```python
# General template for Few shot prompting:

# 1- Explain the task.
# Example1:
# input
# output

# 2- Eg2:
# input
# output

# 3- Eg3:
# input
# output



# Eg: I need help to classify the tweets as positive, negative or neutral

# eg1: "The movie was fantastic"
# output: "Positive"

# eg2: "The movie was a waste of time"
# output: "Negative"

# eg3: "I watched a movie"
# output: "Neutral"


# break - till 6:13, be on time.
```
