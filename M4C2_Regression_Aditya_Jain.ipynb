#Simple Linear Regression

```python
Input: features --> independent
Output: target --> dependent

Find a straight line (linear) that best represents the relationship between the input (features) and the output (target)

best fit line: ML automatically finds the line that has the smallest prediction error. --> best fit line.

y = mx + c

y -> predicted output (predicted house price)
x -> input features
m -> slope
c -> intercept
```

```python
Price = 0.04 × sqft_area + 5
```

```python
import pandas as pd

data = {
    "HouseSize":[800,1000,1200,1500,1800,2000,2200,2500],
    "Price":[32,40,48,60,72,80,88,100]
}

df = pd.DataFrame(data)

df
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["HouseSize"], df["Price"])
```

```python
from sklearn.linear_model import LinearRegression

x = df[["HouseSize"]]
y = df["Price"]


model = LinearRegression()

model.fit(x, y)
```

```python
prediction = model.predict([[1000]])
print(prediction)
```

```python
prediction = model.predict([[1000], [1200], [1500], [1900]])
print(prediction)
```

```python
import numpy as np
np.array([1,2,3,4]).reshape(-1,1)
```

```python
# y = mx + c
```

```python
print(model.coef_)
```

```python
print(model.intercept_)
```

#Multiple Linear Regression

```python
# Limitation of Simple linear regression --> many real-life scenerios depend upon multiple input features
```

```python
The price of the house can dependent upon multiple features:

1. sqft area
2. location
3. bhk
4. age
5. floor
```

```python
# y = w0 + w1*x1 + w2*x2 + ...

# w0 --> intercept
# every feature has its own weight(coefficient)
```

```python
import pandas as pd

data = {

"HouseSize":[1000,1200,1500,1800,2000,2200,2500,2700],

"Bedrooms":[2,2,3,3,4,4,5,5],

"Age":[20,18,15,10,8,6,5,3],

"Price":[40,48,60,72,80,88,100,110]

}

df = pd.DataFrame(data)

df
```

```python
x = df[["HouseSize", "Bedrooms", "Age"]]
y = df["Price"]
```

```python
x
```

```python
y
```

```python
x = df.drop("Price", axis = 1)
x
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
model.predict([[1200]])
```

```python
pred = model.predict([[1200, 3, 10]])
print(pred)
```

```python
# R2 score --> Coefficient of Determination

# - how well the model explains the variation within the data
# - higher is better
```

```python
# MSE --> Mean Squared Error

# error = actual - predicted
# squared error = (actual - predicted)^2
# mean --> avg
# mean squared error = avg of all the squared errors



| Actual | Predicted |
| ------ | --------- |
| 40     | 42        |
| 50     | 49        |
| 60     | 58        |

mse = 9/3 = 3
```

```python
actuals = [40,50,60]
predicted = [42, 49, 58]

from sklearn.metrics import mean_squared_error

mse = mean_squared_error(actuals, predicted)
print(mse)
```
