```python
# Parameters to condider while selecting the model:

# 1. Training performance.
# 2. Test performance
# 3. Bias
# 4. Variance
# 5. Generalization
# 6. Computational Cost
# 7. Business requirements
```

```python
#Bias: it is the error caused by the model being too simple to understand the underlying pattern in the data

# eg: when the relation between x and y is too complex, but we use a very simple model,
# then the model will not be able to capture this relationship well. This will lead to high bias


# For house price prediction, the different important factors are:
# 1. Area
# 2. Location_rank
# 3. Number of bhk
# 4. Age
# 5. Parking + Aminities


# but if model learns: price = location_rank*200
# too simple model leads to high bias
```

```python
# Variance: how much the model's prediction will change when the training data changes

# a high variance model learns the training data very closely --> this leads to overfitting
```

```python
# Condition
# 1. Model too simple:
# - high bias
# - low variance
# - Underfitting


# 2. Model is too complex:
# - low bias
# - high variance
# - overfitting

# We prefer a balanced model --> good generalization

```

```python
# Underfitting and Overfitting

# 1. Underfitting --> the model is unable to learn the important patterns.

# signs:
# training performance --> poor
# test performance --> poor


# 2. Overfitting: model learns the training data too closely --> poor generalization

# signs:
# training performance --> very good
# test performance --> bad


# Good Fit model: that generalizes well

# signs:
# 1. Training performance: 95% accuracy
# 2. test performance: 92% accuracy

```

```python
#Ensemble Models: Ensemble learning combines multiple models to produce a stronger overall prediction
# numerical data --> average
# categorical data --> majority voting

from sklearn.datasets import load_iris

data = load_iris()

x = data.data
y = data.target
```

```python
x[:5]
```

```python
y[:5]
```

```python
data.feature_names
```

```python
data.target_names
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state = 42)
```

```python
# 1. Logistic Regression
# 2. Decision tree classifier

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

lr = LogisticRegression()
dtc = DecisionTreeClassifier(max_depth = 3, random_state = 42)
```

```python
lr.fit(x_train, y_train)
dtc.fit(x_train, y_train)
```

```python
score1 = lr.score(x_test, y_test)
score2 = dtc.score(x_test, y_test)
```
