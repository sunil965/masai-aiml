#Bias

```python
# Bias --> the error caused by the model being too simple or making overly strong assumptions

```

```python
# House price prediction:

# 1. Area
# 2. BHK
# 3. Location
# 4. Age
# 5. Aminities


# but the model simplifies it too much and relies only on the Location feature
# Hence, the model might miss on capturing some of the important relationships

# This leads to high bias
```

```python
# Low  bias --> a flexible model that can capture complex relationships

# too much flexibility in model can lead to high variance
```

#Variance

```python
# variance --> how much of the model's predictions can change when the training data changes
```

```python
# Eg:

# Training data 1 --> Model 1

# Training data 2 --> Model 2

# When Model 1 is very different from Model 2 --> the variance is high


# Variance is the error caused by the model being too sensitive to the particular training
# data
```

```python
# If we train a model on 3 different datasets:

# Dataset1 --> model predicts 90
# Dataset2 --> model predicts 50
# Dataset3 --> model predicts 65

# as the dataset changes, the predictions of the model also changes significantly.
# This is called a high variance model
```

```python
# Variance and Overfitting

# A highly complex model will be very sensitive to:

# outliers, noise, individual datapoints, small changes in the training data


# A complex model --> High variance --> Leads to overfitting
```

#Bias Variance Tradeoff

```python
# Simple model -->
# High Bias, Low Variance
# Leads to Underfitting

# Complex model ->
# Low Bias , High Variance
# Leads to Overfitting


# Prediction error => Bias , Variance, Noise

# High Bias --> Model is too simple
# High Variance --> Model is too Complex (sensitive)

# Overfitting --> model is too complex (memorising the trianing data) --> High variance, Low bias

# Underfitting -> model is too simple --> High bias, low variance
```

```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing()

x = data.data
y = data.target

x[:10]
```

```python
y[:10]
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
                                                      x,y, test_size = 0.2, random_state = 42
                                                      )
```

```python
0.2*20640
```

```python
x_test.shape
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(x_train, y_train)
y_preds = model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_preds)
print(mse)
```

#Cross Validation

```python
# We are too much dependent on the specific test data
# What if, it is not a good representation of the overall dataset
# What if it is very simple as compared to the actual unseen data that it will be later
# test on
```

```python
# # k-fold cross validation
# k = 5

# divide the data into 5 equal parts (fold)
# Fold1
# Fold2
# Fold3
# Fold4
# Fold5


# Now, we will train the model k times, each time, keeping a different fold
# for test and using all the reamining folds for training
```

```python
# Round1:
# Training data --> F2 + F3 + F4 + F5
# Test Data --> F1


# Round2:
# Training data --> F1 + F3 + F4 + F5
# Test Data --> F2


# Round3:
# Training data --> F1 + F2 + F4 + F5
# Test Data --> F3


# Round4:
# Training data --> F1 + F2 + F3 + F5
# Test Data --> F4


# Round5:
# Training data --> F1 + F2 + F3 + F4
# Test Data --> F5
```
