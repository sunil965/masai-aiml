#Supervised vs Unsupervised

```python
# supervised: labeled data
# - linear regression --> regression --> house price prediction
# - logistic regression --> classification --> spam/not spam
# - Decision Trees --> both: classification + regression
# - Random Forest --> both: classification + regression



# unsupervised: unlabeled data
# - k-means
# - other clustering algorithms
```

```python
# k --> number of groups/clusters

# 3 groups, k = 3
# means --> average
```

```python
# k-means uses the average position of points to represent each cluster --> centroid
# centroid --> center of the cluster
```

```python
# step1: select the value of k: k = 3
# step2: select the initial centroids: some random points as the starting centroids
# step3: assign points to the nearest centroid
# step4: re-calcuate the new centroids
# cluster1 --> calculate the new average --> new centroid1
# cluster2 --> calculate the new average --> new centroid2
# cluster3 --> calculate the new average --> new centroid3
```

```python
# nearest --> how close are two datapoints to each other --> distance:

# Eucledian distance:
# A(x1, y1)
# B(x2, y2)


# dist = sqrt((x2-x1)^2 + (y1-y2)^2)


# A(3,2)
# B(7,5)

# dist = sqrt((7-3)^2 + (5-2)^2) = sqrt(16 + 9) = 5
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = np.array([
    [2, 3],
    [3, 2],
    [2, 4],
    [3, 3],
    [8, 9],
    [9, 8],
    [8, 10],
    [10, 9]
])
```

```python
len(data)
```

```python
plt.scatter(data[:,0], data[:,1])
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.title("Scatter Plot")

plt.show()
```

```python
# model = KMeans(n_clusters = 2, random_state = 42)
```

```python
model = KMeans(n_clusters = 2, random_state = 42, n_init= 5)
```

```python
model.fit(data)
```

```python
labels = model.labels_
```

```python
print(labels)
```

```python
print(model.cluster_centers_)
```

```python
plt.scatter(
    data[:,0],
    data[:,1],
    c = labels
)
plt.show()
```

```python
plt.scatter(
    data[:,0],
    data[:,1],
    c = labels
)

plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    marker = "X",
    s = 200
)

plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.title("Scatter Plot")

plt.show()
```

```python
new_data = [[4,5]]
```

```python
pred = model.predict(new_data)
print(pred)
```

```python
new_data = [[9,9]]
pred = model.predict(new_data)
print(pred)
```

```python
new_data = [[4,5], [9,9]]
pred = model.predict(new_data)
print(pred)
```

```python
# Application: Customer Segmentation:  dividing the customers into groups based on similar characteristis or behavioural patterns

import pandas as pd

customers = pd.DataFrame({
    "Customer": ["C1","C2","C3","C4","C5",
                 "C6","C7","C8","C9","C10"],

    "Income": [20,22,25,28,60,65,70,75,40,45],

    "Spending": [20,25,22,30,75,80,78,85,40,45]
})

customers
```

```python
x = customers[["Income", "Spending"]]
print(x)
```

```python
model = KMeans(n_clusters=3, random_state = 42, n_init = 5)
model.fit(x)
```

```python
customers["Cluster"] = model.labels_
```

```python
plt.scatter(customers["Income"], customers["Spending"], c=customers["Cluster"])
plt.show()
```
