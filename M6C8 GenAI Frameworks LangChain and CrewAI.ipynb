#Langchain

```python
# Langchain is a framework that helps the developers to connect the different language models
# to the prompts, external data, tools, etc
```

```python
# Important concepts in Langchain:

# 1- Model: The language model used to generate or interpret the text.
# 2- Prompt: Input/Instructions given to an AI model to generate the response.
# 3- Chain or sequene: steps connected in a well-defined order
# 4 - Tool: a callable function or an external capability
# 5 - Agent: A model-driven system which has the capability of choosing among the available actions.
```

```python
#Prompt: use GCCFE Framework to create a prompt: You need AI to help you solve a user ticket and generate response.

#Chain: eg: Suppose the comapny wants to process the feedback in three steps:

# Customer feedback --> 1. Summarise it --> 2. Classify the issue --> 3. Generate the report
```

#Simple Chain without LLM

```python
def summarize_feedback(feedback):
    return "Feedback received: " + feedback


def classify_sentiment(feedback):
    feedback = feedback.lower()

    if "excellent" in feedback or "good" in feedback:
        return "Positive"

    if "bad" in feedback or "poor" in feedback:
        return "Negative"

    return "Needs review"


def create_report(feedback):
    summary = summarize_feedback(feedback)
    sentiment = classify_sentiment(feedback)

    return {
        "summary": summary,
        "sentiment": sentiment
    }


feedback = "The product quality was excellent."

report = create_report(feedback)

print(report)
```

```python
# crewAI --> team of AI agents: it is a framework designed to coordinatge AI agents that can have different roles
# and work together to solve a larger overall task
```

```python
# Important concepts:

# 1. Role: The responsibility assigned to the AI agent.
# 2 . Agent: An AI agent with a well defined role and responsibilities.
# 3. Goal: The ultimate aim that the AI agent should achieve
# 4. Task: A specific work
# 5. Crew: a team of AI agents working together on the larger task.
```
