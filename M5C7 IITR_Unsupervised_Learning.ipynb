#Intro to Unsupervised Learning

```python
# supervised learning -> labeled data (features + target variable)
# unsupervised learning -> unlabeled data (features)
```

```python
# | Hours Studied | Exam Score |
# | ------------: | ---------: |
# |             2 |         45 |
# |             4 |         60 |
# |             6 |         75 |
# |             8 |         90 |


# labeled data --> Supervised learning



# | Customer | Age | Annual Income |
# | -------- | --: | ------------: |
# | A        |  22 |      3,00,000 |
# | B        |  25 |      3,50,000 |
# | C        |  45 |     12,00,000 |
# | D        |  48 |     15,00,000 |
# | E        |  28 |      4,00,000 |

# there is no target variable --> need to discover the patterns/groups by itself.
# unlabled data --> unsupervised learning
```

```python
# k -> number of groups/clusters

# 3 groups --> k = 3
# means --> average

# k-means: use the average position of the points to represent each cluster --> centroid --> center of the clusters
```

```python
# step1: select the value of k: k = 3
# step2: select the initial centroids: some random points as the starting centroids
# step3: assign points to the nearest centroid
# step4: re-calcuate the new centroids
# cluster1 --> calculate the new average --> new centroid1
# cluster2 --> calculate the new average --> new centroid2
# cluster3 --> calculate the new average --> new centroid3
```

```python
data = np.array([
    [2, 3],
    [3, 2],
    [2, 4],
    [3, 3],
    [8, 9],
    [9, 8],
    [8, 10],
    [10, 9]
])

plt.scatter(data[:,0], data[:,1])
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.title("Scatter Plot")

plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters = 2, random_state = 42)
```

```python
model.fit(data)
```

```python
labels = model.labels_
```

```python
print(labels)
```

```python
plt.scatter(
    data[:,0],
    data[:,1],
    c = labels
)
plt.show()
```

```python
new_data = [[9,9]]
pred = model.predict(new_data)
print(pred)
```
